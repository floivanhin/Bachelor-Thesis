```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import kagglehub
```

Correlation heatmaps including weather data but without cyclically encoded time

```python
path = kagglehub.dataset_download("florianhinrichsen/nyc-taxi-data-2023-24")

PUs_df = pd.read_parquet(f"{path}\hourlyPUs2023-24WEATHER.parquet")
DOs_df = pd.read_parquet(f"{path}\hourlyDOs2023-24WEATHER.parquet")

PUs_df["PULocationID"] = PUs_df["PULocationID"].astype("category")
DOs_df["DOLocationID"] = DOs_df["DOLocationID"].astype("category")

plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.size"] = 15
plt.rcParams["figure.dpi"] = 50
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.colors as mcolors

# ---------------------------------------------------------
# 1. Configuration: Custom Colormap
# ---------------------------------------------------------
# Define a custom linear segmented colormap to highlight negative (red) 
# and positive (yellow) correlations.
# (AI generated)
colors = ["red", "white", "yellow"] 
cmap = mcolors.LinearSegmentedColormap.from_list("red_to_yellow", colors)

# ---------------------------------------------------------
# 2. Data Transformation: Pick-Up (PU) Correlation Matrix
# ---------------------------------------------------------
labelsPUs = [
    "Trip count", "Year", "Month", "Day of the month", "Hour of the day", 
    "Day of the week", "Holiday", "Trip distance (mi)", "Fare amount", 
    "Tip amount", "Total amount paid", "For-hire vehicle %", 
    "Time b/w request and pickup", "Rain (mm)", "Snow depth (m)", 
    "Apparent temperature (°C)", "Snowfall (cm)", "Precipitation(mm)", "Longitude","Latitude"
]


# Calculate the correlation matrix for numerical features and round for display.
PUcorrelation_mtx = PUs_df.corr(numeric_only=True).round(3)

# Reorder the features to position 'tip_amount' at index 9 for consistent visualization.
features = list(PUcorrelation_mtx.columns)
features.remove("tip_amount")
features.insert(9, "tip_amount")
PUcorrelation_mtx = PUcorrelation_mtx.loc[features, features]

# ---------------------------------------------------------
# 3. Visualization: Pick-Up (PU) Correlation Heatmap
# ---------------------------------------------------------
plt.figure(figsize=(32, 14))

# Render the heatmap with annotations and bounds set from -1 to 1.
ax = sns.heatmap(
    PUcorrelation_mtx, 
    annot=True, 
    cmap=cmap, 
    vmin=-1, 
    vmax=1, 
    center=0, 
    linewidths=0.5, 
    linecolor="gray"
)

plt.title("Heatmap of the feature correlations (PU data)", fontsize=50, fontweight="bold", pad=30)
ax.set_xticklabels(labelsPUs, rotation=45, ha="right")
ax.set_yticklabels(labelsPUs, rotation=0)

# plt.savefig("Plots/Correlations/Correlation Heatmaps/PUsWEATHER.png", bbox_inches="tight", dpi=300)
plt.show()

# ---------------------------------------------------------
# 4. Data Transformation: Drop-Off (DO) Correlation Matrix
# ---------------------------------------------------------
labelsDOs = [
    "Trip count", "Year", "Month", "Day of the month", "Hour of the day", 
    "Day of the week", "Holiday", "Trip distance (mi)", "Fare amount", 
    "Tip amount", "Total amount paid", "For-hire vehicle %", 
    "Time b/w pickup and dropoff", "Rain (mm)", "Snow depth (m)", 
    "Apparent temperature (°C)", "Snowfall (cm)", "Precipitation(mm)", "Longitude", "Latitude"
]

# Calculate the correlation matrix for numerical features and round for display.
DOcorrelation_mtx = DOs_df.corr(numeric_only=True).round(3)

# Reorder the features to position 'tip_amount' at index 9.
features = list(DOcorrelation_mtx.columns)
features.remove("tip_amount")
features.insert(9, "tip_amount")
DOcorrelation_mtx = DOcorrelation_mtx.loc[features, features]

# ---------------------------------------------------------
# 5. Visualization: Drop-Off (DO) Correlation Heatmap
# ---------------------------------------------------------
plt.figure(figsize=(32, 14))

# Render the heatmap using identical parameters to the PU matrix for visual consistency.
ax = sns.heatmap(
    DOcorrelation_mtx, 
    annot=True, 
    cmap=cmap, 
    vmin=-1, 
    vmax=1, 
    center=0, 
    linewidths=0.5, 
    linecolor="gray"
)

plt.title("Heatmap of the feature correlations (DO data)", fontsize=50, fontweight="bold", pad=30)
ax.set_xticklabels(labelsDOs, rotation=45, ha="right")
ax.set_yticklabels(labelsDOs, rotation=0)

#plt.savefig("Plots/Correlations/Correlation Heatmaps/DOsWEATHER.pdf", bbox_inches="tight", dpi=300)
plt.show()
```
