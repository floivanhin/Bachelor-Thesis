```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
```

Correlation heatmaps including weather data but without cyclically encoded time

```python
PUs_df = pd.read_parquet("Data\hourlyPUs2023-24WEATHER.parquet")
DOs_df = pd.read_parquet("Data\hourlyDOs2023-24WEATHER.parquet")

PUs_df["PULocationID"] = PUs_df["PULocationID"].astype("category")
DOs_df["DOLocationID"] = DOs_df["DOLocationID"].astype("category")

plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.size"] = 15
```

```python
colors = ["red", "white", "yellow"] 
cmap = mcolors.LinearSegmentedColormap.from_list("red_to_yellow", colors)

labelsPUs=["Trip count", "Year", "Month", "Day of the month", "Hour of the day", "Day of the week", "Holiday", "Trip distance (mi)",
        "Fare amount", "Tip amount", "Total amount paid", "For-hire vevhicle %", "Time b/w request and pickup",
        "Rain (mm)", "Snow depth (m)", "Apparent temperature (°C)", "Snowfall (cm)", "Precipitation(mm)"]

PUcorrelation_mtx = PUs_df.corr(numeric_only=True)
PUcorrelation_mtx = PUcorrelation_mtx.round(3)

features = list(PUcorrelation_mtx.columns)
features.remove("tip_amount")
features.insert(9,"tip_amount")
PUcorrelation_mtx = PUcorrelation_mtx.loc[features,features]

plt.figure(figsize=(32,14),dpi=300)
ax = sns.heatmap(PUcorrelation_mtx, annot=True, cmap=cmap, vmin=-1, vmax=1, center=0, linewidths=0.5, linecolor="gray")
plt.title("Heatmap of the feature correlations (pick-up data)", fontsize=50, fontweight="bold", pad=30)

ax.set_xticklabels(labelsPUs,rotation=45,ha="right")
ax.set_yticklabels(labelsPUs,rotation=0)

plt.savefig("Plots\Correlations\Correlation Heatmaps/PUsWEATHER.png", bbox_inches="tight")

plt.show()

labelsDOs=["Trip count", "Year", "Month", "Day of the month", "Hour of the day", "Day of the week", "Holiday", "Trip distance (mi)",
        "Fare amount", "Tip amount", "Total amount paid", "For-hire vevhicle %", "Time b/w pickup and dropoff",
        "Rain (mm)", "Snow depth (m)", "Apparent temperature (°C)", "Snowfall (cm)", "Precipitation(mm)"]


DOcorrelation_mtx = DOs_df.corr(numeric_only=True)
DOcorrelation_mtx = DOcorrelation_mtx.round(3)

features = list(DOcorrelation_mtx.columns)
features.remove("tip_amount")
features.insert(9,"tip_amount")
DOcorrelation_mtx = DOcorrelation_mtx.loc[features,features]

plt.figure(figsize=(32,14),dpi=300)
ax = sns.heatmap(DOcorrelation_mtx, annot=True, cmap=cmap, vmin=-1, vmax=1, center=0, linewidth=0.5, linecolor="gray")
plt.title("Heatmap of the feature correlations (drop-off data)", fontsize=50, fontweight="bold", pad=30)

ax.set_xticklabels(labelsDOs,rotation=45,ha="right")
ax.set_yticklabels(labelsDOs,rotation=0)

plt.savefig("Plots\Correlations\Correlation Heatmaps/DOsWEATHER.png", bbox_inches="tight")

plt.show()
```
