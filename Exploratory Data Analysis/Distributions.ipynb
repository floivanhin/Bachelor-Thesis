```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.colors as mcolors
import matplotlib.cm as cm
import matplotlib.ticker as ticker
from mpl_toolkits.axes_grid1 import make_axes_locatable
```

```python
PUs_df = pd.read_parquet("Data\hourlyPUs2023-24WEATHER.parquet")
DOs_df = pd.read_parquet("Data\hourlyDOs2023-24WEATHER.parquet")

plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.size"] = 20
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
```

### Trip count

Average trip count by hour

```python
hourly_PUs = PUs_df.groupby("hour")["trip_count"].mean().reset_index()
hourly_DOs = DOs_df.groupby("hour")["trip_count"].mean().reset_index()

plt.figure(figsize=(32,14),dpi=300)

plt.bar(
    hourly_PUs["hour"],
    hourly_PUs["trip_count"],
    color="#C21019"
)

plt.title("Average trip count per hour of the day (PU data)",
          fontsize=50, fontweight="bold",pad=30)
plt.xlabel("Hour of the day", fontweight="bold")
plt.ylabel("Trip count", fontweight="bold")
plt.margins(x=0.01)
plt.xticks(range(0, 24))
plt.grid(axis="y",linestyle="-", alpha=0.7)
plt.gca().set_axisbelow(True)

plt.savefig("Plots\Distributions\TripCount\PUsHourlyTripCount.png")
plt.show()


plt.figure(figsize=(32,14),dpi=300)
plt.bar(
    hourly_DOs["hour"],
    hourly_DOs["trip_count"],
    color="#C21019"
)

plt.title("Average trip count per hour of the day (DO data)",
          fontsize=50, fontweight="bold",pad=30)
plt.xlabel("Hour of the day", fontweight="bold")
plt.ylabel("Trip count", fontweight="bold")
plt.margins(x=0.01)
plt.xticks(range(0, 24))
plt.grid(axis="y",linestyle="-", alpha=0.7)
plt.gca().set_axisbelow(True)

plt.savefig("Plots\Distributions\TripCount\DOsHourlyTripCount.png")
plt.show()
```

Average trip count by day of week

```python
daily_PUs = PUs_df.groupby("DayOfWeek")["trip_count"].mean().reset_index()
daily_DOs = DOs_df.groupby("DayOfWeek")["trip_count"].mean().reset_index()


plt.figure(figsize=(32,14),dpi=300)
plt.bar(
    daily_PUs["DayOfWeek"],
    daily_PUs["trip_count"],
    color="#C21019"
)

plt.title("Average trip count per day of the week (PU data)",
          fontsize=50, fontweight="bold",pad=30)
plt.xlabel("Day of the week", fontweight="bold")
plt.ylabel("Trip count", fontweight="bold")
plt.margins(x=0.01)
plt.grid(axis="y",linestyle="-", alpha=0.7)
plt.gca().set_axisbelow(True)
plt.xticks(range(0, 7),labels=
           ["Mon.","Tue.","Wed.","Thu.","Fri.","Sat.","Sun."])

plt.savefig("Plots\Distributions\TripCount\PUsDailyTripCount.png")
plt.show()


plt.figure(figsize=(32,14),dpi=300)
plt.bar(
    daily_DOs["DayOfWeek"],
    daily_DOs["trip_count"],
    color="#C21019"
)

plt.title("Average trip count per day of the week (DO data)",
          fontsize=50, fontweight="bold",pad=30)
plt.xlabel("Day of the week", fontweight="bold")
plt.ylabel("Trip count", fontweight="bold")
plt.margins(x=0.01)
plt.grid(axis="y",linestyle="-", alpha=0.7)
plt.gca().set_axisbelow(True)
plt.xticks(range(0, 7),labels=
           ["Mon.","Tue.","Wed.","Thu.","Fri.","Sat.","Sun."])

plt.savefig("Plots\Distributions\TripCount\DOsDailyTripCount.png")
plt.show()
```

Trip count by day of week with a seperate bar for holidays

```python
day_mapping = {
    0: "Monday", 1: "Tuesday", 2: "Wednesday",
    3: "Thursday", 4: "Friday", 5: "Saturday", 6: "Sunday"
}

PUs_df["day"] = PUs_df["DayOfWeek"].map(day_mapping)

PUs_df["day_group"] = np.where(PUs_df["is_holiday"] == True, "Holiday", PUs_df["day"])

daily_df = PUs_df.groupby("day_group")["trip_count"].mean().reset_index()

day_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday", "Holiday"]
daily_df["day_group"] = pd.Categorical(daily_df["day_group"], categories=day_order, ordered=True)
grouped_df = daily_df.sort_values("day_group")

plt.figure(figsize=(32,14),dpi=300)
sns.barplot(
    data=daily_df,
    x="day_group",
    y="trip_count",
    color="#C21019" 
)


plt.title("Average trip count per day of the week \nwith a seperate category for holidays (PU data)",
          fontsize=50, fontweight="bold",pad=30)
plt.xlabel("Day of the week or holiday", fontweight="bold")
plt.ylabel("Trip count", fontweight="bold")
plt.margins(x=0.01)
plt.grid(axis="y",linestyle="-", alpha=0.7)
plt.gca().set_axisbelow(True)
plt.xticks(range(0, 8),labels=
           ["Mon.","Tue.","Wed.","Thu.","Fri.","Sat.","Sun.","Holi."])

plt.savefig("Plots\Distributions\TripCount\PUsDailyHolidayTripCount.png")
plt.show()


DOs_df["day"] = DOs_df["DayOfWeek"].map(day_mapping)

DOs_df["day_group"] = np.where(DOs_df["is_holiday"] == True, "Holiday", DOs_df["day"])

daily_df = DOs_df.groupby("day_group")["trip_count"].mean().reset_index()

day_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday", "Holiday"]
daily_df["day_group"] = pd.Categorical(daily_df["day_group"], categories=day_order, ordered=True)
grouped_df = daily_df.sort_values("day_group")

plt.figure(figsize=(32,14),dpi=300)
sns.barplot(
    data=daily_df,
    x="day_group",
    y="trip_count",
    color="#C21019" 
)

plt.title("Average trip count per day of the week \nwith a seperate category for holidays (DO data)",
          fontsize=50, fontweight="bold",pad=30)
plt.xlabel("Day of the week or holiday", fontweight="bold")
plt.ylabel("Trip count", fontweight="bold")
plt.margins(x=0.01)
plt.grid(axis="y",linestyle="-", alpha=0.7)
plt.gca().set_axisbelow(True)
plt.xticks(range(0, 8),labels=
           ["Mon.","Tue.","Wed.","Thu.","Fri.","Sat.","Sun.","Holi."])

plt.savefig("Plots\Distributions\TripCount\DOsDailyHolidayTripCount.png")
plt.show()
```

Heatmap of the trip count by hour and day of week

```python
heatmap_PUs = pd.pivot_table(
    PUs_df,
    values="trip_count",
    index="hour",
    columns="DayOfWeek",
    aggfunc="sum"
)

plt.figure(figsize=(32,14),dpi=300)
ax = sns.heatmap(
    data=heatmap_PUs,
    cmap="flare",
    linewidth=0.5,
    annot=False
)

plt.title("Total trip count by weekday and hour (PU data)",
          fontsize=50,fontweight="bold",pad=30)
ax.set_xlabel("Day of the week", fontweight="bold",labelpad=15)
ax.set_xticklabels(["Mon.","Tue.","Wed.","Thu.","Fri.","Sat.","Sun."])
ax.set_ylabel("Hour of the day", fontweight="bold",labelpad=10)
ax.xaxis.tick_top()
ax.xaxis.set_label_position("top")

cbar = ax.collections[0].colorbar

cbar.ax.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, p: format(int(x), ",")))

plt.savefig("Plots\Distributions\TripCount\PUsDailyHourlyHeatmap.png")
plt.show()


heatmap_DOs = pd.pivot_table(
    DOs_df,
    values="trip_count",
    index="hour",
    columns="DayOfWeek",
    aggfunc="sum"
)

plt.figure(figsize=(32,14),dpi=300)
ax = sns.heatmap(
    data=heatmap_DOs,
    cmap="flare",
    linewidth=0.5,
    annot=False
)

plt.title("Total trip count by weekday and hour (DO data)",
          fontsize=50,fontweight="bold",pad=30)
ax.set_xlabel("Day of the week", fontweight="bold",labelpad=15)
ax.set_xticklabels(["Mon.","Tue.","Wed.","Thu.","Fri.","Sat.","Sun."])
ax.set_ylabel("Hour of the day", fontweight="bold",labelpad=10)
ax.xaxis.tick_top()
ax.xaxis.set_label_position("top")

cbar = ax.collections[0].colorbar

cbar.ax.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, p: format(int(x), ",")))

plt.savefig("Plots\Distributions\TripCount\DOsDailyHourlyHeatmap.png")
plt.show()
```
